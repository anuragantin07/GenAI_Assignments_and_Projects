## Task 1: Basic Function: Price After Discount

```python
def apply_discount(price, discount_percent=5):
    # Ensure discount never exceeds 60% (optional extra)
    if discount_percent > 60:
        print("Warning: Discount percentage exceeds 60%. Capping at 60%.")
        discount_percent = 60

    discount_amount = price * (discount_percent / 100)
    final_price = price - discount_amount
    return final_price

# Test the function
print(f"Price after 10% discount on $1000: ${apply_discount(1000, 10):.2f}")
print(f"Price after default 5% discount on $500: ${apply_discount(500):.2f}")
print(f"Price after 70% discount (capped at 60%) on $200: ${apply_discount(200, 70):.2f}")
```

## Task 2: Recursive Function: Factorial Utility

```python
def factorial(n):
    if n < 0:
        print("Error: Factorial is not defined for negative numbers.")
        return None
    elif n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)

# Test cases
print(f"Factorial of 5: {factorial(5)}")
print(f"Factorial of 0: {factorial(0)}")
print(f"Factorial of -3: {factorial(-3)}")
```

```python
def process_prices(prices):
    # Apply 10% discount to all prices using map() + lambda
    discounted_prices = list(map(lambda price: price * 0.90, prices))

    # Keep only discounted prices that are above 300 using filter() + lambda
    filtered_prices = list(filter(lambda price: price > 300, discounted_prices))

    return discounted_prices, filtered_prices

# Test the function
original_prices_task6 = [100, 500, 900, 50, 750]
disc_prices, filt_prices = process_prices(original_prices_task6)

print(f"Original prices: {original_prices_task6}")
print(f"Prices after 10% discount: {disc_prices}")
print(f"Discounted prices > $300: {filt_prices}")
```

### Task 3: Create a Simple Package (`shop_package`)

This involves creating an `__init__.py` file and two modules: `discount.py` and `billing.py` inside the `shop_package` directory.

```python
%%writefile modules_assignment/shop_package/__init__.py
# This __init__.py makes 'shop_package' a Python package.
# It also allows for direct imports from the package level.

from .discount import apply_discount, flat_discount
from .billing import calculate_total, apply_tax
```

```python
%%writefile modules_assignment/shop_package/discount.py
def apply_discount(price, percent):
    """Applies a percentage discount to a given price."""
    if not isinstance(price, (int, float)) or not isinstance(percent, (int, float)):
        return "Invalid input: price and percent must be numeric."
    if not (0 <= percent <= 100):
        return "Invalid discount percentage: must be between 0 and 100."
    return price - (price * percent / 100)

def flat_discount(price):
    """Applies a flat $50 discount to a price, ensuring price doesn't go below zero."""
    if not isinstance(price, (int, float)):
        return "Invalid input: price must be numeric."
    return max(0, price - 50)
```

```python
%%writefile modules_assignment/shop_package/billing.py
def calculate_total(prices):
    """Calculates the sum of a list of prices."""
    if not isinstance(prices, list):
        return "Invalid input: prices must be a list."
    if not all(isinstance(p, (int, float)) for p in prices):
        return "Invalid input: all prices in the list must be numeric."
    return sum(prices)

def apply_tax(amount):
    """Applies a 5% tax to a given amount."""
    if not isinstance(amount, (int, float)):
        return "Invalid input: amount must be numeric."
    return amount * 1.05 # 5% tax
```

### Task 5: Abstraction (Using Abstract Base Class)

I will define an abstract base class `Payment` with an abstract `process_payment` method. Then, I'll create `CreditCardPayment` and `UPIPayment` subclasses that implement this abstract method.

```python
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def process_payment(self, amount):
        pass

class CreditCardPayment(Payment):
    def process_payment(self, amount):
        print(f"Processing credit card payment of ${amount:.2f}.")

class UPIPayment(Payment):
    def process_payment(self, amount):
        print(f"Processing UPI payment of ${amount:.2f}.")


print("\n--- Task 5: Abstraction (Using Abstract Base Class) ---")
credit_card = CreditCardPayment()
upi = UPIPayment()

credit_card.process_payment(150.75)
upi.process_payment(49.99)
```

```python
import pandas as pd
import numpy as np # Although the restriction says 'NumPy allowed indirectly', it's good practice to import if used by pandas functions internally or for array creation.

# Create a Pandas Series from the list of marks
marks = [78, 85, 90, 66, 72]
marks_series = pd.Series(marks)

print("--- Pandas Series Basics ---")

# Print Series values
print("\nSeries values:")
print(marks_series.values)

# Print Series index
print("\nSeries index:")
print(marks_series.index)

# Print Series data type
print("\nSeries data type:")
print(marks_series.dtype)

# Access first element
print("\nFirst element:", marks_series[0])

# Access last two elements
print("\nLast two elements:")
print(marks_series.tail(2)) # Using .tail() for cleaner display of last elements
```

### Task 2: Mathematical Operations on Series

```python
# Using the marks_series from Task 1
# marks_series is: [78, 85, 90, 66, 72]

print("--- Mathematical Operations on Series ---")

# 1. Add 5 grace marks
marks_add_5 = marks_series + 5
print("\nMarks after adding 5 grace marks:")
print(marks_add_5)

# 2. Subtract 2 marks
marks_subtract_2 = marks_series - 2
print("\nMarks after subtracting 2 marks:")
print(marks_subtract_2)

# 3. Multiply all marks by 1.05
marks_multiply_1_05 = marks_series * 1.05
print("\nMarks after multiplying by 1.05:")
print(marks_multiply_1_05)

# 4. Divide all marks by 2
marks_divide_2 = marks_series / 2
print("\nMarks after dividing by 2:")
print(marks_divide_2)
```

### Task 3: Python Functionalities on Series

```python
# Using the marks_series from Task 1
# marks_series is: [78, 85, 90, 66, 72]

print("--- Python Functionalities on Series ---")

# 1. Find Max, Min, Sum, Mean marks
print("\nMaximum marks:", marks_series.max())
print("Minimum marks:", marks_series.min())
print("Sum of marks:", marks_series.sum())
print("Mean marks:", marks_series.mean())

# 2. Apply a lambda function to check whether each student has passed (> 70)
# The result will be a boolean Series
passed_status = marks_series.apply(lambda x: x > 70)
print("\nPass/Fail status (marks > 70) for each student:")
print(passed_status)

# 3. Count how many students passed
# Summing the boolean Series treats True as 1 and False as 0
num_passed = passed_status.sum()
print(f"\nNumber of students who passed (> 70 marks): {num_passed}")
```

### Task 4: Create a DataFrame

```python
# Data for the DataFrame
students_data = {
    'Name': ['Amit', 'Neha', 'Rahul', 'Sneha', 'Pooja'],
    'Marks': [78, 85, 90, 66, 72],
    'Subject': ['Math', 'Math', 'Science', 'Science', 'Math']
}

print("--- Creating a DataFrame ---")

# 1. Convert it into a DataFrame
students_df = pd.DataFrame(students_data)
print("\nDataFrame created successfully.")

# 2. Print first 3 rows
print("\nFirst 3 rows of the DataFrame:")
print(students_df.head(3))

# 3. Print last 2 rows
print("\nLast 2 rows of the DataFrame:")
print(students_df.tail(2))

# 4. Print DataFrame shape and column names
print("\nDataFrame Shape:", students_df.shape)
print("DataFrame Column Names:", students_df.columns.tolist())
```

### Task 5: Important DataFrame Functions

```python
# Using the students_df DataFrame from Task 4

print("--- Important DataFrame Functions ---")

# 1. Use and print output of info(), describe(), head(), tail()
print("\n--- students_df.info() ---")
students_df.info()

print("\n--- students_df.describe() ---")
print(students_df.describe())

print("\n--- students_df.head() (default 5 rows) ---")
print(students_df.head())

print("\n--- students_df.tail() (default 5 rows) ---")
print(students_df.tail())

# 2. Sort students by Marks in descending order
sorted_students_df = students_df.sort_values(by='Marks', ascending=False)
print("\n--- Students sorted by Marks (descending) ---")
print(sorted_students_df)

# 3. Reset index after sorting
sorted_students_df_reset_index = sorted_students_df.reset_index(drop=True)
print("\n--- Sorted students with reset index ---")
print(sorted_students_df_reset_index)
```

### Task 6: Filtering & Conditional Selection

```python
# Using the students_df DataFrame from Task 4
# students_df:
#     Name  Marks  Subject
# 0   Amit     78     Math
# 1   Neha     85     Math
# 2  Rahul     90  Science
# 3  Sneha     66  Science
# 4  Pooja     72     Math

print("--- Filtering & Conditional Selection ---")

# 1. Select students with Marks greater than 75
students_above_75 = students_df[students_df['Marks'] > 75]
print("\nStudents with Marks > 75:")
print(students_above_75)

# 2. Select students whose Subject is 'Math'
students_math = students_df[students_df['Subject'] == 'Math']
print("\nStudents in 'Math' subject:")
print(students_math)

# 3. Select students with Marks greater than 70 AND in 'Science' subject
students_science_and_marks = students_df[(students_df['Marks'] > 70) & (students_df['Subject'] == 'Science')]
print("\nStudents with Marks > 70 AND in 'Science' subject:")
print(students_science_and_marks)

# 4. Select students whose Name is 'Amit' or 'Pooja'
students_amit_pooja = students_df[students_df['Name'].isin(['Amit', 'Pooja'])]
print("\nStudents named 'Amit' or 'Pooja':")
print(students_amit_pooja)
```

### Task 7: Grouping & Aggregation

```python
# Using the students_df DataFrame from Task 4
# students_df:
#     Name  Marks  Subject
# 0   Amit     78     Math
# 1   Neha     85     Math
# 2  Rahul     90  Science
# 3  Sneha     66  Science
# 4  Pooja     72     Math

print("--- Grouping & Aggregation ---")

# 1. Group by 'Subject' and calculate the average marks for each subject
average_marks_by_subject = students_df.groupby('Subject')['Marks'].mean()
print("\nAverage Marks by Subject:")
print(average_marks_by_subject)

# 2. Count the number of students in each subject
student_count_by_subject = students_df.groupby('Subject')['Name'].count()
print("\nNumber of Students by Subject:")
print(student_count_by_subject)
```

### Task 8: Data Visualization (Basic Plotting)

```python
import matplotlib.pyplot as plt

# Using the average_marks_by_subject Series from Task 7
# average_marks_by_subject:
# Subject
# Math       78.333333
# Science    78.000000
# Name: Marks, dtype: float64

print("--- Data Visualization: Average Marks by Subject ---")

# Create a bar chart of average marks by subject
plt.figure(figsize=(8, 5))
average_marks_by_subject.plot(kind='bar', color=['skyblue', 'lightcoral'])
plt.title('Average Marks by Subject')
plt.xlabel('Subject')
plt.ylabel('Average Marks')
plt.xticks(rotation=0) # Keep subject names horizontal
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()

# Optional: Also visualize the number of students by subject
# student_count_by_subject:
# Subject
# Math       3
# Science    2
# Name: Name, dtype: int64
print("\n--- Data Visualization: Number of Students by Subject ---")
plt.figure(figsize=(8, 5))
student_count_by_subject.plot(kind='bar', color=['lightgreen', 'plum'])
plt.title('Number of Students by Subject')
plt.xlabel('Subject')
plt.ylabel('Number of Students')
plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### Task 9: Handling Missing Values

```python
import numpy as np

# Using the students_df DataFrame from Task 4
# students_df:
#     Name  Marks  Subject
# 0   Amit     78     Math
# 1   Neha     85     Math
# 2  Rahul     90  Science
# 3  Sneha     66  Science
# 4  Pooja     72     Math

print("--- Handling Missing Values ---")

# 1. Introduce some missing values (np.nan)
# Create a copy to avoid modifying the original DataFrame
df_with_nan = students_df.copy()
df_with_nan.loc[0, 'Marks'] = np.nan # Make Amit's marks missing
df_with_nan.loc[2, 'Subject'] = np.nan # Make Rahul's subject missing
print("\nDataFrame with Missing Values:")
print(df_with_nan)

# 2. Check for missing values (isnull(), notnull())
print("\nChecking for null values (df_with_nan.isnull()):")
print(df_with_nan.isnull())
print("\nTotal nulls per column:")
print(df_with_nan.isnull().sum())

# 3. Fill missing values (fillna())
# Fill missing marks with the mean of the 'Marks' column
# Need to calculate mean AFTER setting NaN, otherwise the old mean will be used
mean_marks = df_with_nan['Marks'].mean()
df_filled_marks = df_with_nan.fillna({'Marks': mean_marks})

# Fill missing subjects with a common value, e.g., 'Unknown'
df_filled_all = df_filled_marks.fillna({'Subject': 'Unknown'})
print(f"\nDataFrame after filling missing Marks with mean ({mean_marks:.2f}) and Subject with 'Unknown':")
print(df_filled_all)

# 4. Drop rows with any missing values (dropna())
# Let's re-create df_with_nan to demonstrate dropna cleanly
df_with_nan_for_drop = students_df.copy()
df_with_nan_for_drop.loc[0, 'Marks'] = np.nan
df_with_nan_for_drop.loc[2, 'Subject'] = np.nan

df_dropped_rows = df_with_nan_for_drop.dropna()
print("\nDataFrame after dropping rows with any missing values:")
print(df_dropped_rows)

# Drop columns with any missing values
df_dropped_cols = df_with_nan_for_drop.dropna(axis=1)
print("\nDataFrame after dropping columns with any missing values:")
print(df_dropped_cols)
```

### Task 10: Data Transformation (Adding/Modifying Columns)

```python
# Using the students_df DataFrame from Task 4
# students_df:
#     Name  Marks  Subject
# 0   Amit     78     Math
# 1   Neha     85     Math
# 2  Rahul     90  Science
# 3  Sneha     66  Science
# 4  Pooja     72     Math

print("--- Data Transformation ---")

# 1. Add a new column 'Passed' (True if Marks > 70, False otherwise)
students_df['Passed'] = students_df['Marks'] > 70
print("\nDataFrame after adding 'Passed' column:")
print(students_df)

# 2. Modify an existing column: Scale Marks to be out of 100 (if they were originally out of 120)
# For demonstration, let's assume original marks were out of 100, and we want to scale them to 100 still (no change)
# Or, if we want to simulate a change, we can add a small bonus.

# Let's say we want to add 5 bonus marks to everyone's score, capping at 100.
students_df['Marks_Bonus'] = students_df['Marks'].apply(lambda x: min(x + 5, 100))
print("\nDataFrame after adding 'Marks_Bonus' column (original marks + 5, capped at 100):")
print(students_df)

# Or, if we truly want to modify 'Marks' directly (e.g., scale them if they were originally out of a different total)
# Example: If Marks were originally out of 120, scale to 100.
# Assuming marks are already out of 100, let's just make them integers again if they became float due to NaN handling previously
students_df['Marks'] = students_df['Marks'].astype(int)
print("\nDataFrame with Marks as integers and 'Passed' column:")
print(students_df)
```

## Assignment 11: Matplotlib (Core Plot Types & Visualization)

### Task 1: Line Plot (Sales Trend)

```python
import matplotlib.pyplot as plt

# Sample Sales Data (simulating a Kaggle dataset for demonstration)
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
sales = [150, 180, 220, 200, 250, 280, 300, 290, 320, 350, 380, 400]

print("--- Creating Line Plot for Sales Trend ---")

# Create the line plot
plt.figure(figsize=(10, 6))
plt.plot(months, sales, marker='o', linestyle='-', color='skyblue')

# Add Title and Labels
plt.title('Monthly Sales Trend (Sample Data)')
plt.xlabel('Month')
plt.ylabel('Sales ($ thousands)')

# Add grid for better readability
plt.grid(True, linestyle='--', alpha=0.7)

# Display the plot
plt.show()
```

### Task 2: Scatter Plot (Relationship between two numerical variables)

```python
import matplotlib.pyplot as plt
import numpy as np

# Sample Data for Scatter Plot (simulating a Kaggle dataset for demonstration)
np.random.seed(42) # for reproducibility
x_values = np.random.rand(50) * 100 # 50 random values between 0 and 100
y_values = 2 * x_values + np.random.randn(50) * 20 # A linear relationship with some noise

print("--- Creating Scatter Plot ---")

# Create the scatter plot
plt.figure(figsize=(10, 6))
plt.scatter(x_values, y_values, color='coral', alpha=0.7)

# Add Title and Labels
plt.title('Relationship between Feature X and Feature Y (Sample Data)')
plt.xlabel('Feature X Value')
plt.ylabel('Feature Y Value')

# Add grid for better readability
plt.grid(True, linestyle='--', alpha=0.7)

# Display the plot
plt.show()
```

### Task 3: Bar Plot (Vertical & Horizontal)

```python
import matplotlib.pyplot as plt
import numpy as np

# Sample Data for Bar Plots (simulating a Kaggle dataset for demonstration)
product_categories = ['Electronics', 'Apparel', 'Home & Kitchen', 'Books', 'Sports']
monthly_sales_by_category = [350, 200, 150, 100, 120]

print("--- Creating Vertical Bar Chart ---")

# 1. Create a vertical bar chart
plt.figure(figsize=(10, 6))
plt.bar(product_categories, monthly_sales_by_category, color='teal')

# Add Title and Labels
plt.title('Monthly Sales by Product Category (Vertical Bar Chart)')
plt.xlabel('Product Category')
plt.ylabel('Monthly Sales ($ thousands)')
plt.xticks(rotation=45, ha='right') # Rotate labels for better readability
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout() # Adjust layout to prevent labels from overlapping
plt.show()

print("\n--- Creating Horizontal Bar Chart ---")

# 2. Create a horizontal bar chart
plt.figure(figsize=(10, 6))
plt.barh(product_categories, monthly_sales_by_category, color='lightgreen')

# Add Title and Labels
plt.title('Monthly Sales by Product Category (Horizontal Bar Chart)')
plt.xlabel('Monthly Sales ($ thousands)')
plt.ylabel('Product Category')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### Task 4: Multiple Bar Plot (Comparing Sales of Different Years)

```python
import matplotlib.pyplot as plt
import numpy as np

# Sample Data for Multiple Bar Plot
labels = ['Electronics', 'Apparel', 'Home Goods', 'Books']
sales_2022 = [150, 100, 80, 70]
sales_2023 = [180, 120, 90, 75]
sales_2024 = [200, 130, 100, 80]

x = np.arange(len(labels)) # Label locations
width = 0.2 # Width of the bars

print("--- Creating Multiple Bar Chart ---")

plt.figure(figsize=(12, 7))

# Create bars for each year
plt.bar(x - width, sales_2022, width, label='2022 Sales', color='skyblue')
plt.bar(x, sales_2023, width, label='2023 Sales', color='lightcoral')
plt.bar(x + width, sales_2024, width, label='2024 Sales', color='lightgreen')

# Add Title and Labels
plt.title('Sales Comparison by Category (2022-2024)')
plt.xlabel('Product Category')
plt.ylabel('Sales ($ thousands)')
plt.xticks(x, labels, rotation=45, ha='right') # Set x-axis ticks to category labels

# Add Legend
plt.legend()

# Add grid for better readability
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout() # Adjust layout to prevent labels from overlapping
plt.show()
```

### Task 5: Stacked Bar Chart (Sales by Category and Region)

```python
import matplotlib.pyplot as plt
import numpy as np

# Sample Data for Stacked Bar Chart
categories = ['Electronics', 'Apparel', 'Home Goods', 'Books']
sales_north = np.array([150, 80, 60, 40])
sales_south = np.array([100, 120, 70, 50])
sales_east = np.array([90, 70, 80, 60])

x = np.arange(len(categories)) # Label locations

print("--- Creating Stacked Bar Chart ---")

plt.figure(figsize=(10, 6))

# Plot bars for North region
plt.bar(x, sales_north, label='North Region', color='skyblue')

# Plot bars for South region, stacked on top of North
plt.bar(x, sales_south, bottom=sales_north, label='South Region', color='lightcoral')

# Plot bars for East region, stacked on top of North + South
plt.bar(x, sales_east, bottom=sales_north + sales_south, label='East Region', color='lightgreen')

# Add Title and Labels
plt.title('Regional Sales by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Sales ($ thousands)')
plt.xticks(x, categories, rotation=45, ha='right') # Set x-axis ticks to category labels

# Add Legend
plt.legend()

# Add grid for better readability
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout() # Adjust layout
plt.show()
```

### Task 6: Histogram (Marks Distribution)

```python
import matplotlib.pyplot as plt
import numpy as np

# Sample Data for Histogram (simulating student marks)
np.random.seed(42) # for reproducibility
student_marks = np.random.randint(40, 100, 100) # 100 student marks between 40 and 99

print("--- Creating Histogram for Marks Distribution ---")

plt.figure(figsize=(10, 6))

# Create the histogram
# bins can be chosen to represent grade ranges (e.g., 10-point bins)
plt.hist(student_marks, bins=10, color='purple', edgecolor='black', alpha=0.7)

# Add Title and Labels
plt.title('Distribution of Student Marks')
plt.xlabel('Marks')
plt.ylabel('Number of Students')

# Add grid for better readability
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

### Task 7: Pie Chart (Market Share)

```python
import matplotlib.pyplot as plt

# Sample Data for Pie Chart (simulating market share of different product categories)
categories = ['Electronics', 'Apparel', 'Home & Kitchen', 'Books', 'Others']
market_share = [40, 25, 15, 10, 10] # Percentages
colors = ['gold', 'lightcoral', 'lightskyblue', 'lightgreen', 'plum']
explode = (0.1, 0, 0, 0, 0)  # Explode the 1st slice (Electronics)

print("--- Creating Pie Chart for Market Share ---")

plt.figure(figsize=(8, 8))

# Create the pie chart
plt.pie(market_share, explode=explode, labels=categories, colors=colors,
        autopct='%1.1f%%', shadow=True, startangle=140)

# Add Title
plt.title('Market Share by Product Category')

plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
plt.tight_layout()
plt.show()
```
