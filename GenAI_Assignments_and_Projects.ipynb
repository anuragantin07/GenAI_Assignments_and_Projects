## Task 3: Product Pricing (Dictionaries)

```python
# 1. Create a dictionary price_dict where keys are product names and values are prices.
price_dict = {
    "Laptop": 1200.00,
    "Mouse": 25.50,
    "Keyboard": 75.00,
    "Monitor": 300.00,
    "Webcam": 50.00,
    "Headphones": 100.00,
    "USB Drive": 15.00,
    "External HDD": 80.00
}
print("1. Initial Price Dictionary:", price_dict)

# 2. Write small code blocks to:
# Add a new product with price to price_dict.
print("\n2. Adding 'Printer' (price: 200.00)...")
price_dict["Printer"] = 200.00
print("Price Dictionary after adding Printer:", price_dict)

# Update the price of an existing product
print("\nUpdating price of 'Mouse' to 30.00...")
price_dict["Mouse"] = 30.00
print("Price Dictionary after updating Mouse price:", price_dict)

# Remove a product by name (handle the case when the product does not exist).
product_to_remove_1 = "Webcam"
product_to_remove_2 = "Tablet"

print(f"\nRemoving '{product_to_remove_1}'...")
if product_to_remove_1 in price_dict:
    del price_dict[product_to_remove_1]
    print(f"'{product_to_remove_1}' removed.")
else:
    print(f"'{product_to_remove_1}' not found.")
print("Price Dictionary after removing Webcam:", price_dict)

print(f"\nAttempting to remove '{product_to_remove_2}'...")
if product_to_remove_2 in price_dict:
    del price_dict[product_to_remove_2]
    print(f"'{product_to_remove_2}' removed.")
else:
    print(f"'{product_to_remove_2}' not found.")
print("Price Dictionary after attempting to remove Tablet:", price_dict)

# 3. Print the average price of all products (use only dictionary operations and basic arithmetic).
total_price = sum(price_dict.values())
number_of_products = len(price_dict)

if number_of_products > 0:
    average_price = total_price / number_of_products
    print(f"\n3. Average price of all products: {average_price:.2f}")
else:
    print("\n3. No products in the dictionary to calculate average price.")

# Extra (optional): Print the product with both the maximum and minimum prices.
if price_dict:
    max_price_product = max(price_dict, key=price_dict.get)
    min_price_product = min(price_dict, key=price_dict.get)
    print(f"\nExtra: Product with maximum price: {max_price_product} ({price_dict[max_price_product]:.2f})")
    print(f"Product with minimum price: {min_price_product} ({price_dict[min_price_product]:.2f})")
else:
    print("\nExtra: Cannot find max/min price products as the dictionary is empty.")
```

## Task 4: Combined Operations

```python
# Re-initialize products and product_categories for clarity in this task
products_task4 = [
    "Laptop", "Mouse", "Keyboard", "Monitor", "Headphones", "USB Drive",
    "External HDD", "Printer", "Router", "Webcam"
]

product_categories_task4 = [
    "Electronics", "Electronics", "Electronics", "Electronics", "Electronics", "Storage",
    "Storage", "Peripherals", "Networking", "Electronics"
]

# Ensure price_dict is up-to-date with all products and categories for Task 4
# (Using a copy of price_dict from Task 3 and adding any missing products)
current_price_dict = {
    "Laptop": 1200.00,
    "Mouse": 30.00,      # Updated price from Task 3
    "Keyboard": 75.00,
    "Monitor": 300.00,
    "Headphones": 100.00,
    "USB Drive": 15.00,
    "External HDD": 80.00,
    "Printer": 200.00,    # Added in Task 3
    "Router": 90.00,      # New for Task 4
    "Webcam": 50.00       # Re-added, assuming it wasn't removed permanently for this task
}

# 1. Using the products list and current_price_dict, create a list of tuples named catalog where each tuple is (product_name, price, category).
catalog = []
for i, product_name in enumerate(products_task4):
    price = current_price_dict.get(product_name, 0.0) # Get price, default to 0 if not found
    category = product_categories_task4[i]
    catalog.append((product_name, price, category))

print("1. Product Catalog (list of tuples):")
for item in catalog:
    print(item)


# 2. From catalog, create a new dictionary category_to_products that maps each category to a list of product names in that category
category_to_products = {}
for product_name, _, category in catalog:
    if category not in category_to_products:
        category_to_products[category] = []
    category_to_products[category].append(product_name)

print("\n2. Category to Products Dictionary:")
for category, products_list in category_to_products.items():
    print(f"{category}: {products_list}")


# 3. Print all products that belong to the category that has the maximum number of Products
if category_to_products:
    max_products_category = None
    max_count = 0

    for category, products_list in category_to_products.items():
        current_count = len(products_list)
        if current_count > max_count:
            max_count = current_count
            max_products_category = category

    if max_products_category:
        print(f"\n3. Category with the maximum number of products: '{max_products_category}' (count: {max_count})")
        print(f"Products in '{max_products_category}': {category_to_products[max_products_category]}")
    else:
        print("\n3. No categories found with products.")
else:
    print("\n3. The category_to_products dictionary is empty.")
```
