## Task 1: Product Collections (Lists & Tuples)

```python
# Task 1: Product Collections (Lists & Tuples)

# 1. Create a list named products containing at least 6 product names (strings).
# A list is an ordered, changeable collection of items. Lists are defined with square brackets.
products = [
    "Laptop", "Mouse", "Keyboard", "Monitor", "Webcam", "Headphones",
    "USB Drive", "External HDD"
]
print("1. Initial Products List:", products)

# 2. Create a tuple named sample_product that stores (product_name, price, category) for one product.
# A tuple is an ordered, unchangeable collection of items. Tuples are defined with parentheses.
sample_product = ("Laptop", 1200.00, "Electronics")
print("\n2. Sample Product Tuple:", sample_product)

# 3. Print the 2nd and last product from the products list
# Python lists are 0-indexed, meaning the first item is at index 0, the second at index 1, and so on.
# Negative indexing starts from the end, so -1 is the last item.
print("\n3. The 2nd product (index 1) is:", products[1])
print("The last product (index -1) is:", products[-1])

# 4. Append two new product names to products and then print the updated list.
# The .append() method adds an item to the end of the list.
products.append("Printer")
products.append("Router")
print("\n4. Products List after adding two new products:", products)

# Extra (optional): Convert sample_product into a list, change its price, and convert it back to a tuple.
# Tuples are immutable, so to change an element, we convert it to a list, modify it, then convert it back.
print("\nExtra: Converting tuple to list, changing price, then back to tuple...")
temp_product_list = list(sample_product) # Convert tuple to a list
temp_product_list[1] = 1150.00          # Change the price (which is at index 1)
updated_sample_product_tuple = tuple(temp_product_list) # Convert the list back to a tuple
print("Original sample_product tuple:", sample_product)
print("Updated sample_product tuple with new price:", updated_sample_product_tuple)
```

## Task 2: Categories (Sets)

```python
# Task 2: Categories (Sets)

# For this example, we'll create a simple list of categories that corresponds to our products.
# In a real application, category information might be stored differently.
product_categories = [
    "Electronics", "Electronics", "Electronics", "Electronics", "Electronics", "Electronics",
    "Storage", "Storage", "Peripherals", "Networking"
]

# 1. From your products list, create a set of categories called categories_set.
# A set is an unordered collection of unique items. Duplicates are automatically removed.
categories_set = set(product_categories)
print("1. Unique Categories Set:", categories_set)

# 2. Demonstrate adding a new category to the set and show that duplicates are ignored
# The .add() method adds a single item to the set.
print("\n2. Adding a new category 'Software' to the set...")
categories_set.add("Software")
print("Categories Set after adding 'Software':", categories_set)

print("Attempting to add 'Electronics' again (which is already present)...")
categories_set.add("Electronics") # This item is already in the set, so it will be ignored.
print("Categories Set (duplicates are ignored):", categories_set)

# 3. Show how to check whether a category exists in the set (print a boolean result).
# We can use the 'in' keyword to check for membership in a set.
check_category_1 = "Electronics"
check_category_2 = "Books"
print(f"\n3. Is '{check_category_1}' in the categories set? {check_category_1 in categories_set}")
print(f"Is '{check_category_2}' in the categories set? {check_category_2 in categories_set}")

# Extra (optional): Show how to get the total number of unique categories using a set
# The len() function returns the number of items in a set.
print("\nExtra: Total number of unique categories in the set:", len(categories_set))
```

## Task 3: Product Pricing (Dictionaries)

```python
# Task 3: Product Pricing (Dictionaries)

# 1. Create a dictionary price_dict where keys are product names and values are prices.
# A dictionary is an unordered, changeable collection of key-value pairs. Keys must be unique.
# Dictionaries are defined with curly braces {}.
price_dict = {
    "Laptop": 1200.00,
    "Mouse": 25.50,
    "Keyboard": 75.00,
    "Monitor": 300.00,
    "Webcam": 50.00,
    "Headphones": 100.00,
    "USB Drive": 15.00,
    "External HDD": 80.00
}
print("1. Initial Price Dictionary:", price_dict)

# 2. Write small code blocks to:
# Add a new product with price to price_dict.
# You can add a new key-value pair by assigning a value to a new key.
print("\n2. Adding a new product 'Printer' with price 200.00...")
price_dict["Printer"] = 200.00
print("Price Dictionary after adding Printer:", price_dict)

# Update the price of an existing product
# If a key already exists, assigning a new value to it will update its current value.
print("\nUpdating the price of 'Mouse' to 30.00...")
price_dict["Mouse"] = 30.00
print("Price Dictionary after updating Mouse price:", price_dict)

# Remove a product by name (handle the case when the product does not exist).
# The 'del' keyword can be used to remove a key-value pair.
# It's good practice to check if the key exists first to avoid errors.
product_to_remove_1 = "Webcam"
product_to_remove_2 = "Tablet" # This product does not exist in our dictionary

print(f"\nRemoving product '{product_to_remove_1}'...")
if product_to_remove_1 in price_dict: # Check if the product exists
    del price_dict[product_to_remove_1]
    print(f"'{product_to_remove_1}' was removed.")
else:
    print(f"'{product_to_remove_1}' not found in the dictionary.")
print("Price Dictionary after attempting to remove Webcam:", price_dict)

print(f"\nAttempting to remove non-existent product '{product_to_remove_2}'...")
if product_to_remove_2 in price_dict:
    del price_dict[product_to_remove_2]
    print(f"'{product_to_remove_2}' was removed.")
else:
    print(f"'{product_to_remove_2}' not found in the dictionary.")
print("Price Dictionary after attempting to remove Tablet:", price_dict)

# 3. Print the average price of all products (use only dictionary operations and basic arithmetic).
# We can get all values (prices) using .values() and then sum them up.
total_price = sum(price_dict.values())
number_of_products = len(price_dict) # Get the number of items in the dictionary

if number_of_products > 0:
    average_price = total_price / number_of_products
    print(f"\n3. Average price of all products: {average_price:.2f}")
else:
    print("\n3. No products in the dictionary to calculate average price.")

# Extra (optional): Print the product with both the maximum and minimum prices.
# We can use the max() and min() functions with the 'key' argument to find the key
# (product name) corresponding to the max/min value (price).
if price_dict: # Ensure the dictionary is not empty
    max_price_product_name = max(price_dict, key=price_dict.get) # Finds key with max value
    min_price_product_name = min(price_dict, key=price_dict.get) # Finds key with min value
    print(f"\nExtra: Product with maximum price: {max_price_product_name} (Price: {price_dict[max_price_product_name]:.2f})")
    print(f"Product with minimum price: {min_price_product_name} (Price: {price_dict[min_price_product_name]:.2f})")
else:
    print("\nExtra: Cannot find max/min price products as the dictionary is empty.")
```

## Task 4: Combined Operations

```python
# Task 4: Combined Operations

# For this task, we will use the 'products' list, 'product_categories' list,
# and 'price_dict' that were created and modified in the previous tasks.

# Recall the current state of our data structures:
# products: ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Webcam', 'Headphones', 'USB Drive', 'External HDD', 'Printer', 'Router']
# product_categories: ['Electronics', 'Electronics', 'Electronics', 'Electronics', 'Electronics', 'Electronics', 'Storage', 'Storage', 'Peripherals', 'Networking']
# price_dict: {'Laptop': 1200.0, 'Mouse': 30.0, 'Keyboard': 75.0, 'Monitor': 300.0, 'Headphones': 100.0, 'USB Drive': 15.0, 'External HDD': 80.0, 'Printer': 200.0, 'Router': 90.0}

# Note: For simplicity in this task, we assume the 'products' list from Task 1
# and the 'product_categories' list from Task 2 have the same length and order for mapping.
# The 'price_dict' from Task 3 contains prices for these products.

# 1. Using the 'products' list and 'price_dict', create a list of tuples named catalog
#    where each tuple is (product_name, price, category).
catalog = []
for i, product_name in enumerate(products): # Loop through products list with their index
    # Use .get() to safely retrieve price, returning 0.0 if product not found in price_dict
    price = price_dict.get(product_name, 0.0)
    # Get the corresponding category using the same index 'i'
    category = product_categories[i]
    catalog.append((product_name, price, category))

print("1. Product Catalog (list of tuples - (product_name, price, category)):")
for item in catalog:
    print(item)


# 2. From catalog, create a new dictionary 'category_to_products' that maps each category
#    to a list of product names in that category.
# We'll iterate through the catalog and build this dictionary.
category_to_products = {}
for product_name, _, category in catalog: # Unpack the tuple, ignoring price (_)
    if category not in category_to_products: # If category is not yet a key, add it with an empty list
        category_to_products[category] = []
    category_to_products[category].append(product_name) # Add the product to the category's list

print("\n2. Category to Products Dictionary:")
for category, products_list_in_category in category_to_products.items():
    print(f"  {category}: {products_list_in_category}")


# 3. Print all products that belong to the category that has the maximum number of products.
if category_to_products: # Check if the dictionary is not empty
    max_products_category = None
    max_count = 0

    for category, products_list_in_category in category_to_products.items():
        current_count = len(products_list_in_category)
        if current_count > max_count:
            max_count = current_count
            max_products_category = category

    if max_products_category: # Ensure a category was found (in case all lists were empty)
        print(f"\n3. The category with the maximum number of products is '{max_products_category}' ")
        print(f"   It has {max_count} products. The products are: {category_to_products[max_products_category]}")
    else:
        print("\n3. No categories found with products.")
else:
    print("\n3. The category_to_products dictionary is empty, so no category has products.")
```
